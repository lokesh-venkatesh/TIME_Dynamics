5️⃣ plot_e.py (Immune cell ratios – simulation only)

Definitions:

CD4 = TH1 + TH2

CD8 = TC

Ratios:

CD4/CD8

CD4/Treg

CD8/Treg

Using final time point.

```python
# plot_e.py
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("simulation_results.csv")
last = df.iloc[-1]

CD4 = last["TH1"] + last["TH2"]
CD8 = last["TC"]
Treg = last["Treg"]

ratios = [
    CD4 / CD8,
    CD4 / Treg,
    CD8 / Treg
]

labels = ["CD4/CD8", "CD4/Treg", "CD8/Treg"]

plt.figure(figsize=(6,5))
plt.bar(labels, ratios, color="maroon")

plt.ylabel("Cell Ratio", fontsize=12)
plt.ylim(bottom=0)

plt.tight_layout()
plt.savefig("figure1/figure_1e.png", dpi=300)
plt.close()
```

6️⃣ plot_f.py (Disease vs Cancer comparison – simulation only)

Using:

Disease-free → initial conditions (row 0)

Cancer → final time

Ratios:

CD4/CD8

CD4/Treg

CD8/Treg

TH1/TH2

M1/M2

```python
# plot_f.py
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("simulation_results.csv")

initial = df.iloc[0]
final = df.iloc[-1]

def compute_ratios(row):
    CD4 = row["TH1"] + row["TH2"]
    CD8 = row["TC"]
    Treg = row["Treg"]
    return [
        CD4 / CD8,
        CD4 / Treg,
        CD8 / Treg,
        row["TH1"] / row["TH2"],
        row["M1"] / row["M2"]
    ]

ratios_initial = compute_ratios(initial)
ratios_final = compute_ratios(final)

labels = ["CD4/CD8", "CD4/Treg", "CD8/Treg", "TH1/TH2", "M1/M2"]

x = np.arange(len(labels))
width = 0.35

plt.figure(figsize=(7,5))
plt.bar(x - width/2, ratios_initial, width, label="Disease free", color="green")
plt.bar(x + width/2, ratios_final, width, label="Cancer", color="red")

plt.xticks(x, labels)
plt.ylabel("Cell Ratio", fontsize=12)
plt.legend()
plt.ylim(bottom=0)

plt.tight_layout()
plt.savefig("figure1/figure_1f.png", dpi=300)
plt.close()
```
